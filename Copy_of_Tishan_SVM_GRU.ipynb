## Define Text Preprocessing Function

### Subtask:
Create a function to preprocess natural language queries by lower-casing, removing punctuation, tokenizing, and removing stopwords.


**Reasoning**:
I will start by defining the `preprocess_query` function as outlined in the instructions, incorporating lower-casing, punctuation removal, tokenization, and stopword removal.



```python
import nltk
import re

def preprocess_query(query):
    # Convert to lowercase
    query = query.lower()
    # Remove punctuation and special characters
    query = re.sub(r'[^a-z0-9\s]', '', query)
    # Tokenize the query
    tokens = nltk.word_tokenize(query)
    # Remove stopwords
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

print("Preprocess query function defined.")
```

**Reasoning**:
Now that the necessary date parsing libraries are imported, I will define the `identify_dates` function. This function will use regular expressions to find common date patterns and then leverage `dateutil.parser.parse` to convert identified date strings into datetime objects, returning a list of successfully parsed dates.



```python
def identify_dates(query):
    identified_dates = []
    query_lower = query.lower()

    # Preprocess the query to remove punctuation that might interfere with date parsing
    # but keep spaces, digits, hyphens, and COMMAS
    clean_query = re.sub(r'[^a-z0-9\s,-]', '', query_lower) # Keep hyphens and commas for date formats

    # Regex patterns for common date formats
    # YYYY-MM-DD, YYYY/MM/DD, MM-DD-YYYY, MM/DD/YYYY, Month DD YYYY, DD Month YYYY
    date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b' # DD Month YYYY
        # Removed r'\b\d{4}\b' to prevent ambiguous year-only parsing
    ]

    found_date_strings = set()

    # Extract all potential date strings using regex
    for pattern in date_patterns:
        for match in re.findall(pattern, clean_query):
            found_date_strings.add(match)

    print(f"Found date strings: {found_date_strings}") # Debug print

    # Attempt to parse identified date strings into datetime objects
    for date_str in found_date_strings:
        try:
            # Removed fuzzy=True for more precise date parsing
            parsed_date = parse(date_str)
            identified_dates.append(parsed_date)
            print(f"Parsed '{date_str}' to {parsed_date}") # Debug print
        except ValueError:
            # Handle cases where `dateutil.parser.parse` fails (e.g., non-sensical dates)
            print(f"Could not parse '{date_str}'") # Debug print
            pass

    # Also check for relative dates if relevant (e.g., 'today', 'yesterday', 'tomorrow', 'last year', etc.)
    # This part is more complex and might be added later if needed.

    # Sort the dates to ensure consistency if multiple dates are found
    identified_dates.sort()

    return identified_dates

print("identify_dates function defined.")
```

```python
import pandas as pd
import nltk
import re
import datetime
from google.colab import drive

# --- NLTK Downloads ---
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

# --- Date Parsing Library Check and Install ---
try:
    from dateutil.parser import parse
except ImportError:
    print("Installing python-dateutil...")
    !pip install python-dateutil
    from dateutil.parser import parse

# Ensure Google Drive is mounted for data access
drive.mount('/content/drive', force_remount=True)

# --- Load Data and Initialize Global Variables ---
file_path = '/content/drive/MyDrive/NLP_Tishan/results.csv'
results = pd.read_csv(file_path)

all_teams = set(results['home_team']).union(set(results['away_team']))
all_tournaments = set(results['tournament'])

# --- Function Definitions ---
def preprocess_query(query):
    query = query.lower()
    query = re.sub(r'[^a-z0-9\s]', '', query)
    tokens = nltk.word_tokenize(query)
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

def identify_team_names(query):
    preprocessed_tokens = preprocess_query(query)
    identified_teams_refined = set()
    query_lower = query.lower()
    for team_name in all_teams:
        if team_name.lower() in query_lower:
            identified_teams_refined.add(team_name)
    return list(identified_teams_refined)

def identify_dates(query):
    identified_date_entities = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s/,-]', '', query_lower) # Keep relevant punctuation for date parsing

    # 1. Process explicit specific date formats
    specific_date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b', # DD Month YYYY
    ]

    for pattern in specific_date_patterns:
        for match_str in re.findall(pattern, clean_query):
            try:
                parsed_date = parse(match_str)
                identified_date_entities.append({'type': 'specific', 'date': parsed_date})
            except ValueError:
                pass

    # Keep track of years explicitly identified in ranges to avoid re-processing as standalone
    processed_years_in_range = set()

    # 2. Process year ranges (e.g., "in YYYY", "between YYYY and YYYY")

    # "in YYYY" pattern
    for match in re.findall(r'\b(?:in)\s+(\d{4})\b', clean_query):
        year = int(match)
        if year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    # "between YYYY and YYYY" pattern
    for match in re.findall(r'\b(?:between)\s+(\d{4})\s+(?:and)\s+(\d{4})\b', clean_query):
        year1 = int(match[0])
        year2 = int(match[1])
        start_date = datetime.datetime(year1, 1, 1)
        end_date = datetime.datetime(year2, 12, 31, 23, 59, 59)
        identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
        # Mark all years in the range as processed to avoid standalone processing
        for y in range(year1, year2 + 1):
            processed_years_in_range.add(y)

    # Standalone 4-digit year (treat as a year-long range)
    all_4_digit_numbers_in_query = set(int(y) for y in re.findall(r'\b\d{4}\b', clean_query))

    for year in all_4_digit_numbers_in_query:
        # Check if this year is already part of a specific date or an explicit range
        is_part_of_specific_date = any(
            entity['type'] == 'specific' and entity['date'].year == year
            for entity in identified_date_entities
        )
        if not is_part_of_specific_date and year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    return identified_date_entities

def identify_tournaments(query):
    identified_tournaments = set()
    query_lower = query.lower()
    for tournament_name in all_tournaments:
        if tournament_name.lower() in query_lower:
            identified_tournaments.add(tournament_name)
    return list(identified_tournaments)

def classify_question_type(query):
    query_lower = query.lower()
    if any(keyword in query_lower for keyword in ['won', 'winner', 'who won', 'champion']):
        if any(keyword in query_lower for keyword in ['tournament', 'cup', 'world cup', 'championship']):
            return 'Which team won tournament'
        return 'Who won'
    # Modified to classify 'show me the matches' as 'What was the score'
    elif any(keyword in query_lower for keyword in ['score', 'result', 'what was the score', 'how much', 'show me the matches']):
        return 'What was the score'
    elif any(keyword in query_lower for keyword in ['played', 'match', 'game', 'between']):
        # Keep 'Teams played' for specific queries about who played, but not for 'show me the matches'
        return 'Teams played'
    else:
        return 'General query'

def extract_data(identified_teams, identified_dates, identified_tournaments, question_type, qualifiers=None):
    if qualifiers is None:
        qualifiers = []

    filtered_df = results.copy()
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])

    # --- Step 1: Apply initial filters based on identified entities ---
    if identified_teams:
        team_filter = filtered_df['home_team'].isin(identified_teams) | \
                      filtered_df['away_team'].isin(identified_teams)
        filtered_df = filtered_df[team_filter]

    if identified_dates:
        date_filter_conditions = []
        for date_entity in identified_dates:
            if date_entity['type'] == 'specific':
                date_filter_conditions.append(filtered_df['date'].dt.date == date_entity['date'].date())
            elif date_entity['type'] == 'range':
                start_date = date_entity['start']
                end_date = date_entity['end']
                date_filter_conditions.append(
                    (filtered_df['date'] >= start_date) & (filtered_df['date'] <= end_date)
                )

        if date_filter_conditions:
            combined_date_filter = date_filter_conditions[0]
            for i in range(1, len(date_filter_conditions)):
                combined_date_filter = combined_date_filter | date_filter_conditions[i]
            filtered_df = filtered_df[combined_date_filter]


    if identified_tournaments:
        tournament_filter = filtered_df['tournament'].isin(identified_tournaments)
        filtered_df = filtered_df[tournament_filter]

    # --- Step 2: Apply contextual qualifiers like 'final' ---
    if 'final' in qualifiers and identified_tournaments:
        if not filtered_df.empty:
            # Extract years from identified_dates (only from range types for simplicity in this context for now)
            years_in_query = sorted(list(set(y.year for d_ent in identified_dates if d_ent['type'] == 'range' for y in [d_ent['start'], d_ent['end']])))

            if len(years_in_query) == 1:
                target_year = years_in_query[0]
                filtered_by_year_and_tournament = filtered_df[filtered_df['date'].dt.year == target_year]

                if not filtered_by_year_and_tournament.empty:
                    latest_date_for_final = filtered_by_year_and_tournament.groupby('tournament')['date'].max().reset_index()
                    filtered_df = pd.merge(filtered_by_year_and_tournament, latest_date_for_final, on=['tournament', 'date'])
                else:
                    filtered_df = pd.DataFrame()
            else:
                latest_dates = filtered_df.groupby('tournament')['date'].max().reset_index()
                filtered_df = pd.merge(filtered_df, latest_dates, on=['tournament', 'date'])

    return filtered_df

def generate_answer(filtered_df, question_type, identified_teams=None, identified_dates=None, identified_tournaments=None):
    LARGE_DF_THRESHOLD = 50

    if filtered_df.empty:
        return "No matches found based on your query."

    answer = ""

    if len(filtered_df) > LARGE_DF_THRESHOLD and not (identified_teams or identified_dates or identified_tournaments):
        if question_type == 'Who won':
            return "Your query for 'who won' is too broad. Please specify teams, a tournament, or a date/year for a more precise answer."
        elif question_type == 'What was the score':
            return "Your query for 'what was the score' is too broad. Please specify teams, a tournament, or a date/year for a more precise answer."
        elif question_type == 'Teams played':
            return "Your query for 'teams played' is too broad. Please specify a tournament or a date/year to narrow down the results."


    if question_type == 'Who won':
        if len(filtered_df) == 1:
            match = filtered_df.iloc[0]
            home_score = match['home_score']
            away_score = match['away_score']

            if pd.isna(home_score) or pd.isna(away_score):
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} had no recorded score."
            elif home_score > away_score:
                answer = f"{match['home_team']} won against {match['away_team']} with a score of {int(home_score)}-{int(away_score)} on {match['date'].strftime('%Y-%m-%d')}."
            elif away_score > home_score:
                answer = f"{match['away_team']} won against {match['home_team']} with a score of {int(away_score)}-{int(home_score)} on {match['date'].strftime('%Y-%m-%d')}."
            else:
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} was a draw with a score of {int(home_score)}-{int(away_score)}."
        else:
            team_wins = {}
            for _, match in filtered_df.iterrows():
                home_score = match['home_score']
                away_score = match['away_score']
                home_team = match['home_team']
                away_team = match['away_team']

                if pd.isna(home_score) or pd.isna(away_score):
                    continue

                if home_score > away_score:
                    team_wins[home_team] = team_wins.get(home_team, 0) + 1
                elif away_score > home_score:
                    team_wins[away_team] = team_wins.get(away_team, 0) + 1

            if not team_wins:
                answer = "No clear winner could be determined for these matches (possibly all draws or no scores recorded)."
            else:
                max_wins = 0
                winning_teams = []
                for team, wins in team_wins.items():
                    if wins > max_wins:
                        max_wins = wins
                        winning_teams = [team]
                    elif wins == max_wins:
                        winning_teams.append(team)

                if len(winning_teams) == 1:
                    answer = f"Overall, {winning_teams[0]} had the most wins ({max_wins}) in the queried matches."
                else:
                    answer = f"It was a close contest, with multiple teams having {max_wins} wins: {', '.join(winning_teams)}."

    elif question_type == 'What was the score':
        if len(filtered_df) > LARGE_DF_THRESHOLD:
            return "Too many matches found. Please specify teams, a tournament, or a date/year to get specific scores."
        scores_list = []
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                scores_list.append(f"{match['home_team']} vs {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}: Score not recorded.")
            else:
                scores_list.append(f"{match['home_team']} {int(home_score)}-{int(away_score)} {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}.")
        answer = "\n".join(scores_list)

    elif question_type == 'Which team won tournament':
        team_wins = {}
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            home_team = match['home_team']
            away_team = match['away_team']

            if pd.isna(home_score) or pd.isna(away_score):
                continue

            if home_score > away_score:
                team_wins[home_team] = team_wins.get(home_team, 0) + 1
            elif away_score > home_score:
                team_wins[away_team] = team_wins.get(away_team, 0) + 1

        if not team_wins:
            answer = "No clear winner could be determined for the tournament (possibly all draws or no scores recorded).\nPlease ensure the query specifies a relevant tournament and year if you are looking for a tournament winner."
        else:
            tournament_winner = max(team_wins, key=team_wins.get)
            answer = f"The winner of the tournament based on the available matches is {tournament_winner}."

    elif question_type == 'Teams played':
        if len(filtered_df) > LARGE_DF_THRESHOLD:
             return "Too many teams found. Please specify a tournament or a date/year to narrow down the teams."
        unique_teams = set()
        for _, match in filtered_df.iterrows():
            unique_teams.add(match['home_team'])
            unique_teams.add(match['away_team'])
        answer = f"The teams that played were: {', '.join(sorted(list(unique_teams)))}."

    else:
        answer = "I cannot answer this type of question yet."

    return answer

print("All NLQA helper functions and dependencies defined.")
```

```python
import pandas as pd
import nltk
import re
import datetime
from google.colab import drive

# --- NLTK Downloads ---
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

# --- Date Parsing Library Check and Install ---
try:
    from dateutil.parser import parse
except ImportError:
    print("Installing python-dateutil...")
    !pip install python-dateutil
    from dateutil.parser import parse

# Ensure Google Drive is mounted for data access
drive.mount('/content/drive', force_remount=True)

# --- Load Data and Initialize Global Variables ---
file_path = '/content/drive/MyDrive/NLP_Tishan/results.csv'
results = pd.read_csv(file_path)

all_teams = set(results['home_team']).union(set(results['away_team']))
all_tournaments = set(results['tournament'])

# --- Function Definitions ---
def preprocess_query(query):
    query = query.lower()
    query = re.sub(r'[^a-z0-9\s]', '', query)
    tokens = nltk.word_tokenize(query)
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

def identify_team_names(query):
    preprocessed_tokens = preprocess_query(query)
    identified_teams_refined = set()
    query_lower = query.lower()
    for team_name in all_teams:
        if team_name.lower() in query_lower:
            identified_teams_refined.add(team_name)
    return list(identified_teams_refined)

def identify_dates(query):
    identified_dates = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s,-]', '', query_lower)
    date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b' # DD Month YYYY
    ]
    found_date_strings = set()
    for pattern in date_patterns:
        for match in re.findall(pattern, clean_query):
            found_date_strings.add(match)
    for date_str in found_date_strings:
        try:
            parsed_date = parse(date_str)
            identified_dates.append(parsed_date)
        except ValueError:
            pass
    identified_dates.sort()
    return identified_dates

def identify_tournaments(query):
    identified_tournaments = set()
    query_lower = query.lower()
    for tournament_name in all_tournaments:
        if tournament_name.lower() in query_lower:
            identified_tournaments.add(tournament_name)
    return list(identified_tournaments)

def classify_question_type(query):
    query_lower = query.lower()
    if any(keyword in query_lower for keyword in ['won', 'winner', 'who won', 'champion']):
        if any(keyword in query_lower for keyword in ['tournament', 'cup', 'world cup', 'championship']):
            return 'Which team won tournament'
        return 'Who won'
    elif any(keyword in query_lower for keyword in ['score', 'result', 'what was the score', 'how much']):
        return 'What was the score'
    elif any(keyword in query_lower for keyword in ['played', 'match', 'game', 'between']):
        return 'Teams played'
    else:
        return 'General query'

def extract_data(identified_teams, identified_dates, identified_tournaments, question_type):
    filtered_df = results.copy()
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])
    if identified_teams:
        team_filter = filtered_df['home_team'].isin(identified_teams) | \
                      filtered_df['away_team'].isin(identified_teams)
        filtered_df = filtered_df[team_filter]
    if identified_dates:
        dates_to_filter = [d.date() for d in identified_dates]
        filtered_df = filtered_df[filtered_df['date'].dt.date.isin(dates_to_filter)]
    if identified_tournaments:
        tournament_filter = filtered_df['tournament'].isin(identified_tournaments)
        filtered_df = filtered_df[tournament_filter]
    return filtered_df

def generate_answer(filtered_df, question_type):
    if filtered_df.empty:
        return "No matches found based on your query."
    answer = ""
    if question_type == 'Who won':
        if len(filtered_df) == 1:
            match = filtered_df.iloc[0]
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} had no recorded score."
            elif home_score > away_score:
                answer = f"{match['home_team']} won against {match['away_team']} with a score of {int(home_score)}-{int(away_score)} on {match['date'].strftime('%Y-%m-%d')}."
            elif away_score > home_score:
                answer = f"{match['away_team']} won against {match['home_team']} with a score of {int(away_score)}-{int(home_score)} on {match['date'].strftime('%Y-%m-%d')}."
            else:
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} was a draw with a score of {int(home_score)}-{int(away_score)}."
        else:
            team_wins = {}
            for _, match in filtered_df.iterrows():
                home_score = match['home_score']
                away_score = match['away_score']
                home_team = match['home_team']
                away_team = match['away_team']
                if pd.isna(home_score) or pd.isna(away_score):
                    continue
                if home_score > away_score:
                    team_wins[home_team] = team_wins.get(home_team, 0) + 1
                elif away_score > home_score:
                    team_wins[away_team] = team_wins.get(away_team, 0) + 1
            if not team_wins:
                answer = "No clear winner could be determined for these matches (possibly all draws or no scores recorded). Please refine your query."
            else:
                max_wins = 0
                winning_teams = []
                for team, wins in team_wins.items():
                    if wins > max_wins:
                        max_wins = wins
                        winning_teams = [team]
                    elif wins == max_wins:
                        winning_teams.append(team)
                if len(winning_teams) == 1:
                    answer = f"Overall, {winning_teams[0]} had the most wins ({max_wins}) in the queried matches."
                else:
                    answer = f"It was a close contest, with multiple teams having {max_wins} wins: {', '.join(winning_teams)}."
    elif question_type == 'What was the score':
        scores_list = []
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                scores_list.append(f"{match['home_team']} vs {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}: Score not recorded.")
            else:
                scores_list.append(f"{match['home_team']} {int(home_score)}-{int(away_score)} {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}.")
        answer = "\n".join(scores_list)
    elif question_type == 'Which team won tournament':
        team_wins = {}
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            home_team = match['home_team']
            away_team = match['away_team']
            if pd.isna(home_score) or pd.isna(away_score):
                continue
            if home_score > away_score:
                team_wins[home_team] = team_wins.get(home_team, 0) + 1
            elif away_score > home_score:
                team_wins[away_team] = team_wins.get(away_team, 0) + 1
        if not team_wins:
            answer = "No clear winner could be determined for the tournament (possibly all draws or no scores recorded)."
        else:
            tournament_winner = max(team_wins, key=team_wins.get)
            answer = f"The winner of the tournament based on the available matches is {tournament_winner}."
    elif question_type == 'Teams played':
        unique_teams = set()
        for _, match in filtered_df.iterrows():
            unique_teams.add(match['home_team'])
            unique_teams.add(match['away_team'])
        answer = f"The teams that played were: {', '.join(sorted(list(unique_teams)))}."
    else:
        answer = "I cannot answer this type of question yet."
    return answer

def answer_query(query):
    # 1. Preprocess the query
    # The preprocess_query function is implicitly called within other functions

    # 2. Identify entities
    teams = identify_team_names(query)
    dates = identify_dates(query)
    tournaments = identify_tournaments(query)

    # 3. Classify question type
    question_type = classify_question_type(query)

    # 4. Extract data
    filtered_data = extract_data(teams, dates, tournaments, question_type)

    # 5. Generate answer
    answer = generate_answer(filtered_data, question_type)

    return answer

query_test1 = "Who won the match between Brazil and Germany on 2014-07-08?"
answer_test1 = answer_query(query_test1)
print(f"Query: '{query_test1}'\nAnswer: {answer_test1}\n")
```

def preprocess_query(query):
    query = query.lower()
    query = re.sub(r'[^a-z0-9\s]', '', query)
    tokens = nltk.word_tokenize(query)
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

def identify_team_names(query):

def identify_dates(query):
    identified_date_entities = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s/,-]', '', query_lower) # Keep relevant punctuation for date parsing

    # 1. Process explicit specific date formats
    specific_date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b', # DD Month YYYY
    ]

    for pattern in specific_date_patterns:
        for match_str in re.findall(pattern, clean_query):
            try:
                parsed_date = parse(match_str)
                identified_date_entities.append({'type': 'specific', 'date': parsed_date})
            except ValueError:
                pass

    # Keep track of years explicitly identified in ranges to avoid re-processing as standalone
    processed_years_in_range = set()

    # 2. Process year ranges (e.g., "in YYYY", "between YYYY and YYYY")

    # "in YYYY" pattern
    for match in re.findall(r'\b(?:in)\s+(\d{4})\b', clean_query):
        year = int(match)
        if year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    # "between YYYY and YYYY" pattern
    for match in re.findall(r'\b(?:between)\s+(\d{4})\s+(?:and)\s+(\d{4})\b', clean_query):
        year1 = int(match[0])
        year2 = int(match[1])
        start_date = datetime.datetime(year1, 1, 1)
        end_date = datetime.datetime(year2, 12, 31, 23, 59, 59)
        identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
        # Mark all years in the range as processed to avoid standalone processing
        for y in range(year1, year2 + 1):
            processed_years_in_range.add(y)

    # Standalone 4-digit year (treat as a year-long range)
    all_4_digit_numbers_in_query = set(int(y) for y in re.findall(r'\b\d{4}\b', clean_query))

    for year in all_4_digit_numbers_in_query:
        # Check if this year is already part of a specific date or an explicit range
        is_part_of_specific_date = any(
            entity['type'] == 'specific' and entity['date'].year == year
            for entity in identified_date_entities
        )
        if not is_part_of_specific_date and year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    return identified_date_entities

def classify_question_type(query):
    query_lower = query.lower()
    if any(keyword in query_lower for keyword in ['won', 'winner', 'who won', 'champion']):
        if any(keyword in query_lower for keyword in ['tournament', 'cup', 'world cup', 'championship']):
            return 'Which team won tournament'
        return 'Who won'
    # Modified to classify 'show me the matches' as 'What was the score'
    elif any(keyword in query_lower for keyword in ['score', 'result', 'what was the score', 'how much', 'show me the matches']):
        return 'What was the score'
    elif any(keyword in query_lower for keyword in ['played', 'match', 'game', 'between']):
        # Keep 'Teams played' for specific queries about who played, but not for 'show me the matches'
        return 'Teams played'
    else:
        return 'General query'

def extract_data(identified_teams, identified_dates, identified_tournaments, question_type, qualifiers=None):
    if qualifiers is None:
        qualifiers = []

    filtered_df = results.copy()
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])

    # --- Step 1: Apply initial filters based on identified entities ---
    if identified_teams:
        if len(identified_teams) == 2:
            team1 = identified_teams[0]
            team2 = identified_teams[1]
            # Corrected logic for 'between X and Y' queries
            team_filter = (
                (filtered_df['home_team'] == team1) & (filtered_df['away_team'] == team2)
            ) | (
                (filtered_df['home_team'] == team2) & (filtered_df['away_team'] == team1)
            )
            filtered_df = filtered_df[team_filter]
        else:
            # Existing logic for single team or more than two teams (broad OR condition)
            team_filter = filtered_df['home_team'].isin(identified_teams) | \
                          filtered_df['away_team'].isin(identified_teams)
            filtered_df = filtered_df[team_filter]

    if identified_dates:
        date_filter_conditions = []
        for date_entity in identified_dates:
            if date_entity['type'] == 'specific':
                date_filter_conditions.append(filtered_df['date'].dt.date == date_entity['date'].date())
            elif date_entity['type'] == 'range':
                start_date = date_entity['start']
                end_date = date_entity['end']
                date_filter_conditions.append(
                    (filtered_df['date'] >= start_date) & (filtered_df['date'] <= end_date)
                )

        if date_filter_conditions:
            combined_date_filter = date_filter_conditions[0]
            for i in range(1, len(date_filter_conditions)):
                combined_date_filter = combined_date_filter | date_filter_conditions[i]
            filtered_df = filtered_df[combined_date_filter]


    if identified_tournaments:
        tournament_filter = filtered_df['tournament'].isin(identified_tournaments)
        filtered_df = filtered_df[tournament_filter]

    # --- Step 2: Apply contextual qualifiers like 'final' ---
    if 'final' in qualifiers and identified_tournaments:
        if not filtered_df.empty:
            # Extract years from identified_dates (only from range types for simplicity in this context for now)
            years_in_query = sorted(list(set(y.year for d_ent in identified_dates if d_ent['type'] == 'range' for y in [d_ent['start'], d_ent['end']])))

            if len(years_in_query) == 1:
                target_year = years_in_query[0]
                filtered_by_year_and_tournament = filtered_df[filtered_df['date'].dt.year == target_year]

                if not filtered_by_year_and_tournament.empty:
                    latest_date_for_final = filtered_by_year_and_tournament.groupby('tournament')['date'].max().reset_index()
                    filtered_df = pd.merge(filtered_by_year_and_tournament, latest_date_for_final, on=['tournament', 'date'])
                else:
                    filtered_df = pd.DataFrame() # No matches for that tournament/year
            else:
                latest_dates = filtered_df.groupby('tournament')['date'].max().reset_index()
                filtered_df = pd.merge(filtered_df, latest_dates, on=['tournament', 'date'])

    return filtered_df

## Implement Date Range and Relative Date Parsing

### Subtask:
Extend the `identify_dates` function to support the identification and parsing of date ranges (e.g., 'matches in 2010', 'between 2000 and 2010') and relative date expressions (e.g., 'last week', 'yesterday').


**Reasoning**:
I will modify the `identify_dates` function to parse date ranges like 'in YYYY' and 'between YYYY and YYYY', as well as standalone four-digit years, and represent them as dictionaries with 'start' and 'end' datetime objects. Explicit specific dates will be represented as dictionaries with a 'date' datetime object. This aligns with the subtask instructions to update the function's return to handle a list of dates or a dictionary indicating a date range.



```python
import re
import datetime
from dateutil.parser import parse

def identify_dates(query):
    identified_date_entities = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s/,-]', '', query_lower) # Keep relevant punctuation for date parsing

    # 1. Process explicit specific date formats
    specific_date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b', # DD Month YYYY
    ]

    for pattern in specific_date_patterns:
        for match_str in re.findall(pattern, clean_query):
            try:
                parsed_date = parse(match_str)
                identified_date_entities.append({'type': 'specific', 'date': parsed_date})
            except ValueError:
                pass

    # Keep track of years explicitly identified in ranges to avoid re-processing as standalone
    processed_years_in_range = set()

    # 2. Process year ranges (e.g., "in YYYY", "between YYYY and YYYY")

    # "in YYYY" pattern
    for match in re.findall(r'\b(?:in)\s+(\d{4})\b', clean_query):
        year = int(match)
        if year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    # "between YYYY and YYYY" pattern
    for match in re.findall(r'\b(?:between)\s+(\d{4})\s+(?:and)\s+(\d{4})\b', clean_query):
        year1 = int(match[0])
        year2 = int(match[1])
        start_date = datetime.datetime(year1, 1, 1)
        end_date = datetime.datetime(year2, 12, 31, 23, 59, 59)
        identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
        # Mark all years in the range as processed to avoid standalone processing
        for y in range(year1, year2 + 1):
            processed_years_in_range.add(y)

    # Standalone 4-digit year (treat as a year-long range)
    all_4_digit_numbers_in_query = set(int(y) for y in re.findall(r'\b\d{4}\b', clean_query))

    for year in all_4_digit_numbers_in_query:
        # Check if this year is already part of a specific date or an explicit range
        is_part_of_specific_date = any(
            entity['type'] == 'specific' and entity['date'].year == year
            for entity in identified_date_entities
        )
        if not is_part_of_specific_date and year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    return identified_date_entities

print("identify_dates function updated to handle date ranges and specific dates.")
```

```python
import pandas as pd
import nltk
import re
import datetime
from google.colab import drive

# --- NLTK Downloads ---
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

# --- Date Parsing Library Check and Install ---
try:
    from dateutil.parser import parse
except ImportError:
    print("Installing python-dateutil...")
    !pip install python-dateutil
    from dateutil.parser import parse

# Ensure Google Drive is mounted for data access
drive.mount('/content/drive', force_remount=True)

# --- Load Data and Initialize Global Variables ---
# Base path for CSV files
base_file_path = '/content/drive/MyDrive/NLP_Tishan/'

results = pd.read_csv(base_file_path + 'results.csv')
former_names_df = pd.read_csv(base_file_path + 'former_names.csv')
shootouts_df = pd.read_csv(base_file_path + 'shootouts.csv')
goalscorers_df = pd.read_csv(base_file_path + 'goalscorers.csv')


all_teams = set(results['home_team']).union(set(results['away_team']))
all_tournaments = set(results['tournament'])

# --- Function Definitions ---
def preprocess_query(query):
    query = query.lower()
    query = re.sub(r'[^a-z0-9\s]', '', query)
    tokens = nltk.word_tokenize(query)
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

def identify_team_names(query):
    preprocessed_tokens = preprocess_query(query)
    identified_teams_refined = set()
    query_lower = query.lower()
    for team_name in all_teams:
        if team_name.lower() in query_lower:
            identified_teams_refined.add(team_name)
    return list(identified_teams_refined)

def identify_dates(query):
    identified_date_entities = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s/,-]', '', query_lower) # Keep relevant punctuation for date parsing

    # 1. Process explicit specific date formats
    specific_date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b' # DD Month YYYY
    ]

    for pattern in specific_date_patterns:
        for match_str in re.findall(pattern, clean_query):
            try:
                parsed_date = parse(match_str)
                identified_date_entities.append({'type': 'specific', 'date': parsed_date})
            except ValueError:
                pass

    # Keep track of years explicitly identified in ranges to avoid re-processing as standalone
    processed_years_in_range = set()

    # 2. Process year ranges (e.g., "in YYYY", "between YYYY and YYYY")

    # "in YYYY" pattern
    for match in re.findall(r'\b(?:in)\s+(\d{4})\b', clean_query):
        year = int(match)
        if year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    # "between YYYY and YYYY" pattern
    for match in re.findall(r'\b(?:between)\s+(\d{4})\s+(?:and)\s+(\d{4})\b', clean_query):
        year1 = int(match[0])
        year2 = int(match[1])
        start_date = datetime.datetime(year1, 1, 1)
        end_date = datetime.datetime(year2, 12, 31, 23, 59, 59)
        identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
        # Mark all years in the range as processed to avoid standalone processing
        for y in range(year1, year2 + 1):
            processed_years_in_range.add(y)

    # Standalone 4-digit year (treat as a year-long range)
    all_4_digit_numbers_in_query = set(int(y) for y in re.findall(r'\b\d{4}\b', clean_query))

    for year in all_4_digit_numbers_in_query:
        # Check if this year is already part of a specific date or an explicit range
        is_part_of_specific_date = any(
            entity['type'] == 'specific' and entity['date'].year == year
            for entity in identified_date_entities
        )
        if not is_part_of_specific_date and year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    return identified_date_entities

def identify_tournaments(query):
    identified_tournaments = set()
    query_lower = query.lower()
    for tournament_name in all_tournaments:
        if tournament_name.lower() in query_lower:
            identified_tournaments.add(tournament_name)
    return list(identified_tournaments)

def classify_question_type(query):
    query_lower = query.lower()
    # NEW: Classify 'Top scorer' questions
    if any(keyword in query_lower for keyword in ['most goals', 'top scorer', 'highest scorer']):
        return 'Top scorer'
    elif any(keyword in query_lower for keyword in ['won', 'winner', 'who won', 'champion']):
        if any(keyword in query_lower for keyword in ['tournament', 'cup', 'world cup', 'championship']):
            return 'Which team won tournament'
        return 'Who won'
    elif any(keyword in query_lower for keyword in ['score', 'result', 'what was the score', 'how much', 'show me the matches']):
        return 'What was the score'
    elif any(keyword in query_lower for keyword in ['played', 'match', 'game', 'between']):
        return 'Teams played'
    else:
        return 'General query'

def extract_data(identified_teams, identified_dates, identified_tournaments, question_type, qualifiers=None):
    if qualifiers is None:
        qualifiers = []

    # 'Top scorer' questions do not typically require filtering the 'results' DataFrame initially
    if question_type == 'Top scorer':
        return pd.DataFrame() # Return empty DataFrame as 'goalscorers_df' is used directly in generate_answer

    filtered_df = results.copy()
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])

    # --- Step 1: Apply initial filters based on identified entities ---
    if identified_teams:
        if len(identified_teams) == 2:
            team1 = identified_teams[0]
            team2 = identified_teams[1]
            # Precise AND condition for 'between X and Y' queries
            team_filter = (
                (filtered_df['home_team'] == team1) & (filtered_df['away_team'] == team2)
            ) | (
                (filtered_df['home_team'] == team2) & (filtered_df['away_team'] == team1)
            )
            filtered_df = filtered_df[team_filter]
        else:
            # Existing logic for single team or more than two teams (broad OR condition)
            team_filter = filtered_df['home_team'].isin(identified_teams) | \
                          filtered_df['away_team'].isin(identified_teams)
            filtered_df = filtered_df[team_filter]

    if identified_dates:
        date_filter_conditions = []
        for date_entity in identified_dates:
            if date_entity['type'] == 'specific':
                date_filter_conditions.append(filtered_df['date'].dt.date == date_entity['date'].date())
            elif date_entity['type'] == 'range':
                start_date = date_entity['start']
                end_date = date_entity['end']
                date_filter_conditions.append(
                    (filtered_df['date'] >= start_date) & (filtered_df['date'] <= end_date)
                )

        if date_filter_conditions:
            combined_date_filter = date_filter_conditions[0]
            for i in range(1, len(date_filter_conditions)):
                combined_date_filter = combined_date_filter | date_filter_conditions[i]
            filtered_df = filtered_df[combined_date_filter]


    if identified_tournaments:
        tournament_filter = filtered_df['tournament'].isin(identified_tournaments)
        filtered_df = filtered_df[tournament_filter]

    # --- Step 2: Apply contextual qualifiers like 'final' ---
    if 'final' in qualifiers and identified_tournaments:
        if not filtered_df.empty:
            # Extract years from identified_dates (only from range types for simplicity in this context for now)
            years_in_query = sorted(list(set(y.year for d_ent in identified_dates if d_ent['type'] == 'range' for y in [d_ent['start'], d_ent['end']])))

            if len(years_in_query) == 1:
                target_year = years_in_query[0]
                filtered_by_year_and_tournament = filtered_df[filtered_df['date'].dt.year == target_year]

                if not filtered_by_year_and_tournament.empty:
                    latest_date_for_final = filtered_by_year_and_tournament.groupby('tournament')['date'].max().reset_index()
                    filtered_df = pd.merge(filtered_by_year_and_tournament, latest_date_for_final, on=['tournament', 'date'])
                else:
                    filtered_df = pd.DataFrame() # No matches for that tournament/year
            else:
                latest_dates = filtered_df.groupby('tournament')['date'].max().reset_index()
                filtered_df = pd.merge(filtered_df, latest_dates, on=['tournament', 'date'])

    return filtered_df

def generate_answer(filtered_df, question_type, identified_teams=None, identified_dates=None, identified_tournaments=None):
    LARGE_DF_THRESHOLD = 50

    if filtered_df.empty and question_type != 'Top scorer':
        return "No matches found based on your query."

    answer = ""

    # General broadness check: Only return broad message if no entities were identified AND df is large
    # This check won't apply to 'Top scorer' as it doesn't filter `results` initially
    if len(filtered_df) > LARGE_DF_THRESHOLD and not (identified_teams or identified_dates or identified_tournaments):
        if question_type == 'Who won':
            return "Your query for 'who won' is too broad. Please specify teams, a tournament, or a date/year for a more precise answer."
        elif question_type == 'What was the score':
            return "Your query for 'what was the score' is too broad. Please specify teams, a tournament, or a date/year for a more precise answer."
        elif question_type == 'Teams played':
            return "Your query for 'teams played' is too broad. Please specify a tournament or a date/year to narrow down the results."


    if question_type == 'Who won':
        if len(filtered_df) == 1:
            match = filtered_df.iloc[0]
            home_score = match['home_score']
            away_score = match['away_score']

            if pd.isna(home_score) or pd.isna(away_score):
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} had no recorded score."
            elif home_score > away_score:
                answer = f"{match['home_team']} won against {match['away_team']} with a score of {int(home_score)}-{int(away_score)} on {match['date'].strftime('%Y-%m-%d')}."
            elif away_score > home_score:
                answer = f"{match['away_team']} won against {match['home_team']} with a score of {int(away_score)}-{int(home_score)} on {match['date'].strftime('%Y-%m-%d')}."
            else:
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} was a draw with a score of {int(home_score)}-{int(away_score)}."
        else:
            team_wins = {}
            for _, match in filtered_df.iterrows():
                home_score = match['home_score']
                away_score = match['away_score']
                home_team = match['home_team']
                away_team = match['away_team']

                if pd.isna(home_score) or pd.isna(away_score):
                    continue # Skip matches with missing scores

                if home_score > away_score:
                    team_wins[home_team] = team_wins.get(home_team, 0) + 1
                elif away_score > home_score:
                    team_wins[away_team] = team_wins.get(away_team, 0) + 1

            if not team_wins:
                answer = "No clear winner could be determined for these matches (possibly all draws or no scores recorded)."
            else:
                max_wins = 0
                winning_teams = []
                for team, wins in team_wins.items():
                    if wins > max_wins:
                        max_wins = wins
                        winning_teams = [team]
                    elif wins == max_wins:
                        winning_teams.append(team)

                if len(winning_teams) == 1:
                    answer = f"Overall, {winning_teams[0]} had the most wins ({max_wins}) in the queried matches."
                else:
                    answer = f"It was a close contest, with multiple teams having {max_wins} wins: {', '.join(winning_teams)}."

    elif question_type == 'What was the score':
        scores_list = []
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                scores_list.append(f"{match['home_team']} vs {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}: Score not recorded.")
            else:
                scores_list.append(f"{match['home_team']} {int(home_score)}-{int(away_score)} {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}.")
        answer = "\n".join(scores_list)

    elif question_type == 'Which team won tournament':
        team_wins = {}
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            home_team = match['home_team']
            away_team = match['away_team']

            if pd.isna(home_score) or pd.isna(away_score):
                continue # Skip matches with missing scores

            if home_score > away_score:
                team_wins[home_team] = team_wins.get(home_team, 0) + 1
            elif away_score > home_score:
                team_wins[away_team] = team_wins.get(away_team, 0) + 1

        if not team_wins:
            answer = "No clear winner could be determined for the tournament (possibly all draws or no scores recorded).\nPlease ensure the query specifies a relevant tournament and year if you are looking for a tournament winner."
        else:
            tournament_winner = max(team_wins, key=team_wins.get)
            answer = f"The winner of the tournament based on the available matches is {tournament_winner}."

    elif question_type == 'Teams played':
        unique_teams = set()
        for _, match in filtered_df.iterrows():
            unique_teams.add(match['home_team'])
            unique_teams.add(match['away_team'])
        answer = f"The teams that played were: {', '.join(sorted(list(unique_teams)))}."

    # NEW: Handle 'Top scorer' questions
    elif question_type == 'Top scorer':
        if 'goalscorers_df' in globals() and not goalscorers_df.empty:
            top_scorers = goalscorers_df['scorer'].value_counts().head(10)
            if not top_scorers.empty:
                answer = "Top 10 Goalscorers (all time):\n"
                for scorer, count in top_scorers.items():
                    answer += f"- {scorer}: {count} goals\n"
            else:
                answer = "Could not determine top scorers from the available data."
        else:
            answer = "I'm sorry, I cannot provide information on individual top goalscorers at this moment as the goalscorers data is not loaded or is empty."

    else:
        answer = "I cannot answer this type of question yet."

    return answer

def answer_query(query):
    # 1. Preprocess the query (implicitly called within other functions)

    # 2. Identify entities
    teams = identify_team_names(query)
    dates = identify_dates(query)
    tournaments = identify_tournaments(query)

    # Identify qualifiers like 'final'
    qualifiers = []
    if 'final' in query.lower():
        qualifiers.append('final')

    # 3. Classify question type
    question_type = classify_question_type(query)

    # 4. Extract data, passing the identified qualifiers and structured dates
    # For 'Top scorer', filtered_data will be an empty DataFrame as it's not needed for initial filtering
    filtered_data = extract_data(teams, dates, tournaments, question_type, qualifiers)

    # 5. Generate answer, passing identified entities for ambiguity handling
    answer = generate_answer(filtered_data, question_type, identified_teams=teams, identified_dates=dates, identified_tournaments=tournaments)

    return answer

print("All NLQA helper functions and dependencies updated.")
```

```python
import pandas as pd
import nltk
import re
import datetime
from google.colab import drive

# --- NLTK Downloads ---
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

# --- Date Parsing Library Check and Install ---
try:
    from dateutil.parser import parse
except ImportError:
    print("Installing python-dateutil...")
    !pip install python-dateutil
    from dateutil.parser import parse

# Ensure Google Drive is mounted for data access
drive.mount('/content/drive', force_remount=True)

# --- Load Data and Initialize Global Variables ---
# Base path for CSV files
base_file_path = '/content/drive/MyDrive/NLP_Tishan/'

results = pd.read_csv(base_file_path + 'results.csv')
former_names_df = pd.read_csv(base_file_path + 'former_names.csv')
shootouts_df = pd.read_csv(base_file_path + 'shootouts.csv')
goalscorers_df = pd.read_csv(base_file_path + 'goalscorers.csv')


all_teams = set(results['home_team']).union(set(results['away_team']))
all_tournaments = set(results['tournament'])

# --- Function Definitions ---
def preprocess_query(query):
    query = query.lower()
    query = re.sub(r'[^a-z0-9\s]', '', query)
    tokens = nltk.word_tokenize(query)
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

def identify_team_names(query):
    preprocessed_tokens = preprocess_query(query)
    identified_teams_refined = set()
    query_lower = query.lower()
    for team_name in all_teams:
        if team_name.lower() in query_lower:
            identified_teams_refined.add(team_name)
    return list(identified_teams_refined)

def identify_dates(query):
    identified_date_entities = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s/,-]', '', query_lower) # Keep relevant punctuation for date parsing

    # 1. Process explicit specific date formats
    specific_date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b', # DD Month YYYY
    ]

    for pattern in specific_date_patterns:
        for match_str in re.findall(pattern, clean_query):
            try:
                parsed_date = parse(match_str)
                identified_date_entities.append({'type': 'specific', 'date': parsed_date})
            except ValueError:
                pass

    # Keep track of years explicitly identified in ranges to avoid re-processing as standalone
    processed_years_in_range = set()

    # 2. Process year ranges (e.g., "in YYYY", "between YYYY and YYYY")

    # "in YYYY" pattern
    for match in re.findall(r'\b(?:in)\s+(\d{4})\b', clean_query):
        year = int(match)
        if year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    # "between YYYY and YYYY" pattern
    for match in re.findall(r'\b(?:between)\s+(\d{4})\s+(?:and)\s+(\d{4})\b', clean_query):
        year1 = int(match[0])
        year2 = int(match[1])
        start_date = datetime.datetime(year1, 1, 1)
        end_date = datetime.datetime(year2, 12, 31, 23, 59, 59)
        identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
        # Mark all years in the range as processed to avoid standalone processing
        for y in range(year1, year2 + 1):
            processed_years_in_range.add(y)

    # Standalone 4-digit year (treat as a year-long range)
    all_4_digit_numbers_in_query = set(int(y) for y in re.findall(r'\b\d{4}\b', clean_query))

    for year in all_4_digit_numbers_in_query:
        # Check if this year is already part of a specific date or an explicit range
        is_part_of_specific_date = any(
            entity['type'] == 'specific' and entity['date'].year == year
            for entity in identified_date_entities
        )
        if not is_part_of_specific_date and year not in processed_years_in_range:
            start_date = datetime.datetime(year, 1, 1)
            end_date = datetime.datetime(year, 12, 31, 23, 59, 59)
            identified_date_entities.append({'type': 'range', 'start': start_date, 'end': end_date})
            processed_years_in_range.add(year)

    return identified_date_entities

def identify_tournaments(query):
    identified_tournaments = set()
    query_lower = query.lower()
    for tournament_name in all_tournaments:
        if tournament_name.lower() in query_lower:
            identified_tournaments.add(tournament_name)
    return list(identified_tournaments)

def classify_question_type(query):
    query_lower = query.lower()
    if any(keyword in query_lower for keyword in ['won', 'winner', 'who won', 'champion']):
        if any(keyword in query_lower for keyword in ['tournament', 'cup', 'world cup', 'championship']):
            return 'Which team won tournament'
        return 'Who won'
    # Modified to classify 'show me the matches' as 'What was the score'
    elif any(keyword in query_lower for keyword in ['score', 'result', 'what was the score', 'how much', 'show me the matches']):
        return 'What was the score'
    elif any(keyword in query_lower for keyword in ['played', 'match', 'game', 'between']):
        # Keep 'Teams played' for specific queries about who played, but not for 'show me the matches'
        return 'Teams played'
    else:
        return 'General query'

def extract_data(identified_teams, identified_dates, identified_tournaments, question_type, qualifiers=None):
    if qualifiers is None:
        qualifiers = []

    filtered_df = results.copy()
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])

    # --- Step 1: Apply initial filters based on identified entities ---
    if identified_teams:
        if len(identified_teams) == 2:
            team1 = identified_teams[0]
            team2 = identified_teams[1]
            # Precise AND condition for 'between X and Y' queries
            team_filter = (
                (filtered_df['home_team'] == team1) & (filtered_df['away_team'] == team2)
            ) | (
                (filtered_df['home_team'] == team2) & (filtered_df['away_team'] == team1)
            )
            filtered_df = filtered_df[team_filter]
        else:
            # Existing logic for single team or more than two teams (broad OR condition)
            team_filter = filtered_df['home_team'].isin(identified_teams) | \
                          filtered_df['away_team'].isin(identified_teams)
            filtered_df = filtered_df[team_filter]

    if identified_dates:
        date_filter_conditions = []
        for date_entity in identified_dates:
            if date_entity['type'] == 'specific':
                date_filter_conditions.append(filtered_df['date'].dt.date == date_entity['date'].date())
            elif date_entity['type'] == 'range':
                start_date = date_entity['start']
                end_date = date_entity['end']
                date_filter_conditions.append(
                    (filtered_df['date'] >= start_date) & (filtered_df['date'] <= end_date)
                )

        if date_filter_conditions:
            combined_date_filter = date_filter_conditions[0]
            for i in range(1, len(date_filter_conditions)):
                combined_date_filter = combined_date_filter | date_filter_conditions[i]
            filtered_df = filtered_df[combined_date_filter]


    if identified_tournaments:
        tournament_filter = filtered_df['tournament'].isin(identified_tournaments)
        filtered_df = filtered_df[tournament_filter]

    # --- Step 2: Apply contextual qualifiers like 'final' ---
    if 'final' in qualifiers and identified_tournaments:
        if not filtered_df.empty:
            # Extract years from identified_dates (only from range types for simplicity in this context for now)
            years_in_query = sorted(list(set(y.year for d_ent in identified_dates if d_ent['type'] == 'range' for y in [d_ent['start'], d_ent['end']])))

            if len(years_in_query) == 1:
                target_year = years_in_query[0]
                filtered_by_year_and_tournament = filtered_df[filtered_df['date'].dt.year == target_year]

                if not filtered_by_year_and_tournament.empty:
                    latest_date_for_final = filtered_by_year_and_tournament.groupby('tournament')['date'].max().reset_index()
                    filtered_df = pd.merge(filtered_by_year_and_tournament, latest_date_for_final, on=['tournament', 'date'])
                else:
                    filtered_df = pd.DataFrame() # No matches for that tournament/year
            else:
                latest_dates = filtered_df.groupby('tournament')['date'].max().reset_index()
                filtered_df = pd.merge(filtered_df, latest_dates, on=['tournament', 'date'])

    return filtered_df

def generate_answer(filtered_df, question_type, identified_teams=None, identified_dates=None, identified_tournaments=None):
    LARGE_DF_THRESHOLD = 50

    if filtered_df.empty:
        return "No matches found based on your query."

    answer = ""

    # General broadness check: Only return broad message if no entities were identified AND df is large
    if len(filtered_df) > LARGE_DF_THRESHOLD and not (identified_teams or identified_dates or identified_tournaments):
        if question_type == 'Who won':
            return "Your query for 'who won' is too broad. Please specify teams, a tournament, or a date/year for a more precise answer."
        elif question_type == 'What was the score':
            return "Your query for 'what was the score' is too broad. Please specify teams, a tournament, or a date/year for a more precise answer."
        elif question_type == 'Teams played':
            return "Your query for 'teams played' is too broad. Please specify a tournament or a date/year to narrow down the results."


    if question_type == 'Who won':
        if len(filtered_df) == 1:
            match = filtered_df.iloc[0]
            home_score = match['home_score']
            away_score = match['away_score']

            if pd.isna(home_score) or pd.isna(away_score):
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} had no recorded score."
            elif home_score > away_score:
                answer = f"{match['home_team']} won against {match['away_team']} with a score of {int(home_score)}-{int(away_score)} on {match['date'].strftime('%Y-%m-%d')}."
            elif away_score > home_score:
                answer = f"{match['away_team']} won against {match['home_team']} with a score of {int(away_score)}-{int(home_score)} on {match['date'].strftime('%Y-%m-%d')}."
            else:
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} was a draw with a score of {int(home_score)}-{int(away_score)}."
        else:
            team_wins = {}
            for _, match in filtered_df.iterrows():
                home_score = match['home_score']
                away_score = match['away_score']
                home_team = match['home_team']
                away_team = match['away_team']

                if pd.isna(home_score) or pd.isna(away_score):
                    continue # Skip matches with missing scores

                if home_score > away_score:
                    team_wins[home_team] = team_wins.get(home_team, 0) + 1
                elif away_score > home_score:
                    team_wins[away_team] = team_wins.get(away_team, 0) + 1

            if not team_wins:
                answer = "No clear winner could be determined for these matches (possibly all draws or no scores recorded)."
            else:
                max_wins = 0
                winning_teams = []
                for team, wins in team_wins.items():
                    if wins > max_wins:
                        max_wins = wins
                        winning_teams = [team]
                    elif wins == max_wins:
                        winning_teams.append(team)

                if len(winning_teams) == 1:
                    answer = f"Overall, {winning_teams[0]} had the most wins ({max_wins}) in the queried matches."
                else:
                    answer = f"It was a close contest, with multiple teams having {max_wins} wins: {', '.join(winning_teams)}."

    elif question_type == 'What was the score':
        scores_list = []
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                scores_list.append(f"{match['home_team']} vs {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}: Score not recorded.")
            else:
                scores_list.append(f"{match['home_team']} {int(home_score)}-{int(away_score)} {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}.")
        answer = "\n".join(scores_list)

    elif question_type == 'Which team won tournament':
        team_wins = {}
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            home_team = match['home_team']
            away_team = match['away_team']

            if pd.isna(home_score) or pd.isna(away_score):
                continue # Skip matches with missing scores

            if home_score > away_score:
                team_wins[home_team] = team_wins.get(home_team, 0) + 1
            elif away_score > home_score:
                team_wins[away_team] = team_wins.get(away_team, 0) + 1

        if not team_wins:
            answer = "No clear winner could be determined for the tournament (possibly all draws or no scores recorded).\nPlease ensure the query specifies a relevant tournament and year if you are looking for a tournament winner."
        else:
            tournament_winner = max(team_wins, key=team_wins.get)
            answer = f"The winner of the tournament based on the available matches is {tournament_winner}."

    elif question_type == 'Teams played':
        unique_teams = set()
        for _, match in filtered_df.iterrows():
            unique_teams.add(match['home_team'])
            unique_teams.add(match['away_team'])
        answer = f"The teams that played were: {', '.join(sorted(list(unique_teams)))}."

    else:
        answer = "I cannot answer this type of question yet."

    return answer

def answer_query(query):
    # 1. Preprocess the query (implicitly called within other functions)

    # 2. Identify entities
    teams = identify_team_names(query)
    dates = identify_dates(query)
    tournaments = identify_tournaments(query)

    # Identify qualifiers like 'final'
    qualifiers = []
    if 'final' in query.lower():
        qualifiers.append('final')

    # 3. Classify question type
    question_type = classify_question_type(query)

    # 4. Extract data, passing the identified qualifiers and structured dates
    filtered_data = extract_data(teams, dates, tournaments, question_type, qualifiers)

    # 5. Generate answer, passing identified entities for ambiguity handling
    answer = generate_answer(filtered_data, question_type, identified_teams=teams, identified_dates=dates, identified_tournaments=tournaments)

    return answer

print("All NLQA helper functions and dependencies updated with 'between X and Y' logic.")
```
