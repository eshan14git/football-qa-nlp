## Define Text Preprocessing Function

### Subtask:
Create a function to preprocess natural language queries by lower-casing, removing punctuation, tokenizing, and removing stopwords.


**Reasoning**:
I will start by defining the `preprocess_query` function as outlined in the instructions, incorporating lower-casing, punctuation removal, tokenization, and stopword removal.



```python
import nltk
import re

def preprocess_query(query):
    # Convert to lowercase
    query = query.lower()
    # Remove punctuation and special characters
    query = re.sub(r'[^a-z0-9\s]', '', query)
    # Tokenize the query
    tokens = nltk.word_tokenize(query)
    # Remove stopwords
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

print("Preprocess query function defined.")
```

**Reasoning**:
Now that the necessary date parsing libraries are imported, I will define the `identify_dates` function. This function will use regular expressions to find common date patterns and then leverage `dateutil.parser.parse` to convert identified date strings into datetime objects, returning a list of successfully parsed dates.



```python
def identify_dates(query):
    identified_dates = []
    query_lower = query.lower()

    # Preprocess the query to remove punctuation that might interfere with date parsing
    # but keep spaces, digits, hyphens, and COMMAS
    clean_query = re.sub(r'[^a-z0-9\s,-]', '', query_lower) # Keep hyphens and commas for date formats

    # Regex patterns for common date formats
    # YYYY-MM-DD, YYYY/MM/DD, MM-DD-YYYY, MM/DD/YYYY, Month DD YYYY, DD Month YYYY
    date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b' # DD Month YYYY
        # Removed r'\b\d{4}\b' to prevent ambiguous year-only parsing
    ]

    found_date_strings = set()

    # Extract all potential date strings using regex
    for pattern in date_patterns:
        for match in re.findall(pattern, clean_query):
            found_date_strings.add(match)

    print(f"Found date strings: {found_date_strings}") # Debug print

    # Attempt to parse identified date strings into datetime objects
    for date_str in found_date_strings:
        try:
            # Removed fuzzy=True for more precise date parsing
            parsed_date = parse(date_str)
            identified_dates.append(parsed_date)
            print(f"Parsed '{date_str}' to {parsed_date}") # Debug print
        except ValueError:
            # Handle cases where `dateutil.parser.parse` fails (e.g., non-sensical dates)
            print(f"Could not parse '{date_str}'") # Debug print
            pass

    # Also check for relative dates if relevant (e.g., 'today', 'yesterday', 'tomorrow', 'last year', etc.)
    # This part is more complex and might be added later if needed.

    # Sort the dates to ensure consistency if multiple dates are found
    identified_dates.sort()

    return identified_dates

print("identify_dates function defined.")
```

```python
def generate_answer(filtered_df, question_type):
    if filtered_df.empty:
        return "No matches found based on your query."

    answer = ""

    if question_type == 'Who won':
        if len(filtered_df) == 1:
            # Handle a single match
            match = filtered_df.iloc[0]
            home_score = match['home_score']
            away_score = match['away_score']

            if pd.isna(home_score) or pd.isna(away_score):
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} had no recorded score."
            elif home_score > away_score:
                answer = f"{match['home_team']} won against {match['away_team']} with a score of {int(home_score)}-{int(away_score)} on {match['date'].strftime('%Y-%m-%d')}."
            elif away_score > home_score:
                answer = f"{match['away_team']} won against {match['home_team']} with a score of {int(away_score)}-{int(home_score)} on {match['date'].strftime('%Y-%m-%d')}."
            else:
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} was a draw with a score of {int(home_score)}-{int(away_score)}."
        else:
            # Handle multiple matches (head-to-head or general 'who won' for a set of teams)
            team_wins = {}
            for _, match in filtered_df.iterrows():
                home_score = match['home_score']
                away_score = match['away_score']
                home_team = match['home_team']
                away_team = match['away_team']

                if pd.isna(home_score) or pd.isna(away_score):
                    continue # Skip matches with missing scores

                if home_score > away_score:
                    team_wins[home_team] = team_wins.get(home_team, 0) + 1
                elif away_score > home_score:
                    team_wins[away_team] = team_wins.get(away_team, 0) + 1

            if not team_wins:
                answer = "No clear winner could be determined for these matches (possibly all draws or no scores recorded)."
            else:
                max_wins = 0
                winning_teams = []
                for team, wins in team_wins.items():
                    if wins > max_wins:
                        max_wins = wins
                        winning_teams = [team]
                    elif wins == max_wins:
                        winning_teams.append(team)

                if len(winning_teams) == 1:
                    answer = f"Overall, {winning_teams[0]} had the most wins ({max_wins}) in the queried matches."
                else:
                    answer = f"It was a close contest, with multiple teams having {max_wins} wins: {', '.join(winning_teams)}."

    elif question_type == 'What was the score':
        scores_list = []
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                scores_list.append(f"{match['home_team']} vs {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}: Score not recorded.")
            else:
                scores_list.append(f"{match['home_team']} {int(home_score)}-{int(away_score)} {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}.")
        answer = "\n".join(scores_list)

    elif question_type == 'Which team won tournament':
        # Assuming that `filtered_df` contains matches from a single tournament or relevant matches
        # A simplistic approach: team with the most wins in the filtered data is the 'winner'
        # For more complex tournament structures, this would need a more sophisticated logic
        # (e.g., identifying final match, checking specific stages).
        team_wins = {}
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            home_team = match['home_team']
            away_team = match['away_team']

            if pd.isna(home_score) or pd.isna(away_score):
                continue # Skip matches with missing scores

            if home_score > away_score:
                team_wins[home_team] = team_wins.get(home_team, 0) + 1
            elif away_score > home_score:
                team_wins[away_team] = team_wins.get(away_team, 0) + 1

        if not team_wins:
            answer = "No clear winner could be determined for the tournament (possibly all draws or no scores recorded)."
        else:
            tournament_winner = max(team_wins, key=team_wins.get)
            answer = f"The winner of the tournament based on the available matches is {tournament_winner}."

    elif question_type == 'Teams played':
        unique_teams = set()
        for _, match in filtered_df.iterrows():
            unique_teams.add(match['home_team'])
            unique_teams.add(match['away_team'])
        answer = f"The teams that played were: {', '.join(sorted(list(unique_teams)))}."

    else:
        answer = "I cannot answer this type of question yet."

    return answer

print("generate_answer function defined.")
```

```python
import pandas as pd
import nltk
import re
import datetime
from google.colab import drive

# --- NLTK Downloads ---
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

# --- Date Parsing Library Check and Install ---
try:
    from dateutil.parser import parse
except ImportError:
    print("Installing python-dateutil...")
    !pip install python-dateutil
    from dateutil.parser import parse

# Ensure Google Drive is mounted for data access
drive.mount('/content/drive', force_remount=True)

# --- Load Data and Initialize Global Variables ---
file_path = '/content/drive/MyDrive/NLP_Tishan/results.csv'
results = pd.read_csv(file_path)

all_teams = set(results['home_team']).union(set(results['away_team']))
all_tournaments = set(results['tournament'])

# --- Function Definitions ---
def preprocess_query(query):
    query = query.lower()
    query = re.sub(r'[^a-z0-9\s]', '', query)
    tokens = nltk.word_tokenize(query)
    stop_words = set(nltk.corpus.stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

def identify_team_names(query):
    preprocessed_tokens = preprocess_query(query)
    identified_teams_refined = set()
    query_lower = query.lower()
    for team_name in all_teams:
        if team_name.lower() in query_lower:
            identified_teams_refined.add(team_name)
    return list(identified_teams_refined)

def identify_dates(query):
    identified_dates = []
    query_lower = query.lower()
    clean_query = re.sub(r'[^a-z0-9\s,-]', '', query_lower)
    date_patterns = [
        r'\b\d{4}[-/]\d{2}[-/]\d{2}\b', # YYYY-MM-DD or YYYY/MM/DD
        r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', # MM-DD-YYYY or MM/DD/YYYY
        r'\b(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}\b', # Month DD, YYYY
        r'\b\d{1,2}\s+(?:january|february|march|april|may|june|july|august|september|october|november|december|jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{4}\b' # DD Month YYYY
    ]
    found_date_strings = set()
    for pattern in date_patterns:
        for match in re.findall(pattern, clean_query):
            found_date_strings.add(match)
    for date_str in found_date_strings:
        try:
            parsed_date = parse(date_str)
            identified_dates.append(parsed_date)
        except ValueError:
            pass
    identified_dates.sort()
    return identified_dates

def identify_tournaments(query):
    identified_tournaments = set()
    query_lower = query.lower()
    for tournament_name in all_tournaments:
        if tournament_name.lower() in query_lower:
            identified_tournaments.add(tournament_name)
    return list(identified_tournaments)

def classify_question_type(query):
    query_lower = query.lower()
    if any(keyword in query_lower for keyword in ['won', 'winner', 'who won', 'champion']):
        if any(keyword in query_lower for keyword in ['tournament', 'cup', 'world cup', 'championship']):
            return 'Which team won tournament'
        return 'Who won'
    elif any(keyword in query_lower for keyword in ['score', 'result', 'what was the score', 'how much']):
        return 'What was the score'
    elif any(keyword in query_lower for keyword in ['played', 'match', 'game', 'between']):
        return 'Teams played'
    else:
        return 'General query'

def extract_data(identified_teams, identified_dates, identified_tournaments, question_type):
    filtered_df = results.copy()
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])
    if identified_teams:
        team_filter = filtered_df['home_team'].isin(identified_teams) | \
                      filtered_df['away_team'].isin(identified_teams)
        filtered_df = filtered_df[team_filter]
    if identified_dates:
        dates_to_filter = [d.date() for d in identified_dates]
        filtered_df = filtered_df[filtered_df['date'].dt.date.isin(dates_to_filter)]
    if identified_tournaments:
        tournament_filter = filtered_df['tournament'].isin(identified_tournaments)
        filtered_df = filtered_df[tournament_filter]
    return filtered_df

def generate_answer(filtered_df, question_type):
    if filtered_df.empty:
        return "No matches found based on your query."
    answer = ""
    if question_type == 'Who won':
        if len(filtered_df) == 1:
            match = filtered_df.iloc[0]
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} had no recorded score."
            elif home_score > away_score:
                answer = f"{match['home_team']} won against {match['away_team']} with a score of {int(home_score)}-{int(away_score)} on {match['date'].strftime('%Y-%m-%d')}."
            elif away_score > home_score:
                answer = f"{match['away_team']} won against {match['home_team']} with a score of {int(away_score)}-{int(home_score)} on {match['date'].strftime('%Y-%m-%d')}."
            else:
                answer = f"The match between {match['home_team']} and {match['away_team']} on {match['date'].strftime('%Y-%m-%d')} was a draw with a score of {int(home_score)}-{int(away_score)}."
        else:
            team_wins = {}
            for _, match in filtered_df.iterrows():
                home_score = match['home_score']
                away_score = match['away_score']
                home_team = match['home_team']
                away_team = match['away_team']
                if pd.isna(home_score) or pd.isna(away_score):
                    continue
                if home_score > away_score:
                    team_wins[home_team] = team_wins.get(home_team, 0) + 1
                elif away_score > home_score:
                    team_wins[away_team] = team_wins.get(away_team, 0) + 1
            if not team_wins:
                answer = "No clear winner could be determined for these matches (possibly all draws or no scores recorded). Please refine your query."
            else:
                max_wins = 0
                winning_teams = []
                for team, wins in team_wins.items():
                    if wins > max_wins:
                        max_wins = wins
                        winning_teams = [team]
                    elif wins == max_wins:
                        winning_teams.append(team)
                if len(winning_teams) == 1:
                    answer = f"Overall, {winning_teams[0]} had the most wins ({max_wins}) in the queried matches."
                else:
                    answer = f"It was a close contest, with multiple teams having {max_wins} wins: {', '.join(winning_teams)}."
    elif question_type == 'What was the score':
        scores_list = []
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            if pd.isna(home_score) or pd.isna(away_score):
                scores_list.append(f"{match['home_team']} vs {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}: Score not recorded.")
            else:
                scores_list.append(f"{match['home_team']} {int(home_score)}-{int(away_score)} {match['away_team']} on {match['date'].strftime('%Y-%m-%d')}.")
        answer = "\n".join(scores_list)
    elif question_type == 'Which team won tournament':
        team_wins = {}
        for _, match in filtered_df.iterrows():
            home_score = match['home_score']
            away_score = match['away_score']
            home_team = match['home_team']
            away_team = match['away_team']
            if pd.isna(home_score) or pd.isna(away_score):
                continue
            if home_score > away_score:
                team_wins[home_team] = team_wins.get(home_team, 0) + 1
            elif away_score > home_score:
                team_wins[away_team] = team_wins.get(away_team, 0) + 1
        if not team_wins:
            answer = "No clear winner could be determined for the tournament (possibly all draws or no scores recorded)."
        else:
            tournament_winner = max(team_wins, key=team_wins.get)
            answer = f"The winner of the tournament based on the available matches is {tournament_winner}."
    elif question_type == 'Teams played':
        unique_teams = set()
        for _, match in filtered_df.iterrows():
            unique_teams.add(match['home_team'])
            unique_teams.add(match['away_team'])
        answer = f"The teams that played were: {', '.join(sorted(list(unique_teams)))}."
    else:
        answer = "I cannot answer this type of question yet."
    return answer

def answer_query(query):
    # 1. Preprocess the query
    # The preprocess_query function is implicitly called within other functions

    # 2. Identify entities
    teams = identify_team_names(query)
    dates = identify_dates(query)
    tournaments = identify_tournaments(query)

    # 3. Classify question type
    question_type = classify_question_type(query)

    # 4. Extract data
    filtered_data = extract_data(teams, dates, tournaments, question_type)

    # 5. Generate answer
    answer = generate_answer(filtered_data, question_type)

    return answer

query_test1 = "Who won the match between Brazil and Germany on 2014-07-08?"
answer_test1 = answer_query(query_test1)
print(f"Query: '{query_test1}'\nAnswer: {answer_test1}\n")
```
