## 3. Missing Values Analysis

```python
# Check for missing values in each dataset

datasets = {
    "Results Dataset": results_df,
    "Goalscorers Dataset": scorers_df,
    "Shootouts Dataset": shootouts_df,
    "Former Names Dataset": former_names_df
}

for dataset_name, dataset in datasets.items():
    print(f"\n=== {dataset_name} ===")
    print(dataset.isnull().sum())
```

## 4. Duplicate Records Analysis

```python
# Check for duplicate records in each dataset

for dataset_name, dataset in datasets.items():
    duplicate_count = dataset.duplicated().sum()
    print(f"{dataset_name}: {duplicate_count} duplicate records")
```

## 5. Inspect Duplicate Records

```python
# Display duplicate rows in the Goalscorers dataset

duplicate_scorers = scorers_df[scorers_df.duplicated(keep=False)]

print(f"Total duplicate rows: {len(duplicate_scorers)}")

duplicate_scorers.sort_values(
    by=["date", "home_team", "away_team", "scorer"]
).head(20)
```

## 7. Dataset Relationships

The Football Question Answering Assistant uses four related datasets.

- **Results Dataset** is the primary dataset containing international match information such as teams, scores, tournament, venue, and match date.

- **Goalscorers Dataset** provides detailed information about individual goals scored in matches. It can be linked to the Results dataset using the match date, home team, and away team.

- **Shootouts Dataset** contains information about matches decided by penalty shootouts. It also links to the Results dataset using the match date, home team, and away team.

- **Former Names Dataset** is a reference table used to map historical country names to their current names. This helps maintain consistency across datasets during preprocessing.
