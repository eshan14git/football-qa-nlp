### Duplicate Record Analysis

The Goalscorers dataset contains 82 duplicate-looking records (128 rows involved in duplicate groups).

These records were manually investigated by comparing them with the corresponding match results in the Results dataset. The investigation showed that the repeated scorer entries accurately reflected the total number of goals scored in historical matches where detailed information such as the goal minute was unavailable.

Since these records preserve valid historical scoring information rather than representing accidental data duplication, they were retained in the dataset.

```python
# Duplicate analysis summary

duplicate_count = scorers_clean.duplicated().sum()

print(f"Duplicate records identified: {duplicate_count}")
print("Decision: Duplicate-looking records were retained after validation.")
```

```python
# Check missing values in each cleaned dataset

print("Results Dataset")
print(results_clean.isnull().sum())

print("\nGoalscorers Dataset")
print(scorers_clean.isnull().sum())

print("\nShootouts Dataset")
print(shootouts_clean.isnull().sum())

print("\nFormer Names Dataset")
print(former_names_clean.isnull().sum())
```

```python
# Display records with missing scorer names

missing_scorer = scorers_clean[scorers_clean["scorer"].isnull()]

missing_scorer
```

```python
# Replace missing scorer names with "Unknown"

scorers_clean["scorer"] = scorers_clean["scorer"].fillna("Unknown")
```

```python
# Verify missing scorer values have been handled

print("Missing scorer values:", scorers_clean["scorer"].isnull().sum())

print("\nUnique replacement value:")

print(scorers_clean[scorers_clean["scorer"] == "Unknown"].head())
```

### Handling Missing Goal Minutes

The `minute` column contains 256 missing values. These represent historical matches where the exact goal minute was unavailable in the original data source.

Since the missing values reflect unavailable historical information rather than data entry errors, they were retained as `NaN`. No imputation was performed because assigning artificial values (such as the mean, median, or a placeholder minute) could introduce misleading information into the dataset.

```python
# Verify remaining missing values in the minute column

print("Missing goal minutes:", scorers_clean["minute"].isnull().sum())
```

```python
# Display records with missing first_shooter values

missing_first_shooter = shootouts_clean[shootouts_clean["first_shooter"].isnull()]

missing_first_shooter
```

```python
# Replace missing first shooter names with "Unknown"

shootouts_clean["first_shooter"] = shootouts_clean["first_shooter"].fillna("Unknown")
```

```python
# Verify missing first_shooter values have been handled

print("Missing first_shooter values:", shootouts_clean["first_shooter"].isnull().sum())

print("\nSample records with replacement value:")

print(shootouts_clean[shootouts_clean["first_shooter"] == "Unknown"].head())
```

```python
# Display all historical country name mappings

former_names_clean
```

```python
# Check which former country names appear in the Results dataset

former_names = set(former_names_clean["former"])

results_former = (
    set(results_clean["home_team"])
    | set(results_clean["away_team"])
)

matches = former_names.intersection(results_former)

print("Former country names found in Results dataset:")
print(sorted(matches))
```

```python
# Check former country names in the Goalscorers dataset

scorers_teams = (
    set(scorers_clean["home_team"])
    | set(scorers_clean["away_team"])
    | set(scorers_clean["team"])
)

scorers_matches = former_names.intersection(scorers_teams)

print("Former country names found in Goalscorers dataset:")
print(sorted(scorers_matches))


# Check former country names in the Shootouts dataset

shootouts_teams = (
    set(shootouts_clean["home_team"])
    | set(shootouts_clean["away_team"])
)

shootouts_matches = former_names.intersection(shootouts_teams)

print("\nFormer country names found in Shootouts dataset:")
print(sorted(shootouts_matches))
```

```python
# Final validation of the cleaned datasets

datasets = {
    "Results": results_clean,
    "Goalscorers": scorers_clean,
    "Shootouts": shootouts_clean,
    "Former Names": former_names_clean
}

for name, dataset in datasets.items():
    print(f"\n{name} Dataset")
    print("-" * 40)
    print("Shape:", dataset.shape)
    print("\nMissing Values:")
    print(dataset.isnull().sum())
    print("\nDuplicate Records:", dataset.duplicated().sum())
    print("\nData Types:")
    print(dataset.dtypes)
```

### Country Name Normalization

The `former_names.csv` reference table was examined to identify historical country names requiring normalization.

A comparison between the reference table and the Results, Goalscorers, and Shootouts datasets showed that none of the historical country names were present. This indicates that the datasets already use standardized country names.

Therefore, no country name normalization was required during preprocessing.

```python
results_clean.to_csv("football-qa-nlp/data/results_clean.csv", index=False)
scorers_clean.to_csv("football-qa-nlp/data/goalscorers_clean.csv", index=False)
shootouts_clean.to_csv("football-qa-nlp/data/shootouts_clean.csv", index=False)
former_names_clean.to_csv("football-qa-nlp/data/former_names_clean.csv", index=False)
```
